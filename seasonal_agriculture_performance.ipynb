```python
df.isnull().sum().sort_values(ascending=False).head(10)
```

```python
numeric_columns=df.select_dtypes(include=np.number).columns
for col in numeric_columns:
  if df[col].isnull().sum()>0:
    df[col]=df[col].fillna(df[col].median())
print("REMAINING MISSING VALUES:",df.isnull().sum().sum())
```

### Missing Value Treatment

Missing numerical values were replaced using the median of the
corresponding column. Median imputation was selected because it is less
affected by extreme values and allows the available observations to be
retained for further analysis.

#**Outlier Analysis**

```python
outlier_summary=[]
for col in numeric_columns:
  q1=df[col].quantile(0.25)
  q3=df[col].quantile(0.75)
  iqr=q3-q1
  lower=q1-1.5*iqr
  upper=q3+1.5*iqr
  count=((df[col]<lower)| (df[col]>upper)).sum()
  outlier_summary.append({
      "Column":col,
      "Lower_Bound":lower,
      "Upper_Bound":upper,
      "Outlier_Count":count,
      "Outlier_Percentage":round(count/len(df)*100,2)
  })
outlier_summary=pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count",ascending=False
)
outlier_summary
```

# **Statistical Test — Does Yield Differ Across Seasons?**

```python
from scipy.stats import f_oneway
kharif = df[
    df["Season"] == "Kharif"
]["Yield_Tonnes_Ha"]
rabi = df[
    df["Season"] == "Rabi"
]["Yield_Tonnes_Ha"]
zaid = df[
    df["Season"] == "Zaid"
]["Yield_Tonnes_Ha"]
f_stat, p_value = f_oneway(
    kharif,
    rabi,
    zaid
)
print("F-statistic:", f_stat)
print("P-value:", p_value)
if p_value < 0.05:
    print(
        "Yield differs significantly across seasons."
    )
else:
    print(
        "No statistically significant difference in yield across seasons."
    )
```

### Statistical Interpretation

A one-way ANOVA test was used to examine whether the mean agricultural
yield differs across Kharif, Rabi and Zaid seasons.

A p-value below 0.05 indicates that the observed seasonal differences
in mean yield are statistically significant.

# Limitations

1. The analysis is based only on the variables available in the dataset.

2. Observed relationships do not necessarily imply causation.

3. Seasonal differences may also be influenced by crop type, location,

4. Missing numerical observations were handled using median imputation,

5. The analysis provides evidence from the available dataset and should
   not be treated as a universal conclusion for all agricultural regions.
